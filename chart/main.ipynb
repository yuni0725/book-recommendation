```python
import pandas as pd

df = pd.read_csv("data.csv")
df = df[df['연도'] == 2023]

df['사서교사']
```

```python
df = df.dropna(subset=['사서교사'])

df['사서교사'] = df['사서교사'].astype(int)
df['사서교사']
```

```python
df['사서교사'].value_counts()
```

```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.figure(figsize=(10, 6))
plt.xlabel('사서교사 수')
plt.ylabel('학교 수')
plt.title('전국 초중고 교내 사서교사 수')

plt.bar(df['사서교사'].value_counts().index.astype(int), df['사서교사'].value_counts().values)

# Add text labels with '명' suffix
for i, v in enumerate(df['사서교사'].value_counts().values):
    plt.text(df['사서교사'].value_counts().index[i], v, f'{v}', 
             ha='center', va='bottom')

# Add '명' suffix to x-axis labels
plt.xticks(df['사서교사'].value_counts().index.astype(int), 
           [f'{x}명' for x in df['사서교사'].value_counts().index.astype(int)])


plt.savefig('file.png', dpi=600)
plt.tight_layout()
plt.show()
```
