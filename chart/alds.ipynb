```python
import pandas as pd

df = pd.read_csv('data.csv')
```

```python
df2024 = df[df['연도'] == 2024]
```

```python
elementary = df2024[df2024['학교급'] == '초등학교']
middle = df2024[df2024['학교급'] == '중학교']
high = df2024[df2024['학교급'] == '고등학교']

elementary = elementary.dropna(subset=['사서교사'])
middle = middle.dropna(subset=['사서교사'])
high = high.dropna(subset=['사서교사'])
```
