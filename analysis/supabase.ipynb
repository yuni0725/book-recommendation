```python
df = pd.read_csv("data3 copy.csv")

def parse_pg_array(s):
    s = s.strip().replace("[", "").replace("]", "") # remove outer quotes and {}
    return [item.strip("'") for item in s.split(', ') if item]

df['category'] = df['category'].apply(parse_pg_array)
```

```python
df = df.dropna(subset=['isbn'])

df = df.where(pd.notnull(df), None)
```

```python
# Reset book_id starting from 9990
df['book_id'] = range(10017, 10017 + len(df))
df['book_id'] = df['book_id'].astype(str)

df['book_id']
```

```python
# 업로드
for record in df.to_dict(orient="records"):
    supabase.table("books").insert(record).execute()
```
