```python
raw['카테고리'] = raw['카테고리'].astype(object)

for index, row in raw.iterrows():
    df_row = df[df['자료명'] == row['자료명']]
    try:
        raw.at[index, '썸네일'] = df_row['카카오썸네일'].values[0]
        raw.at[index, '책소개'] = df_row['책소개'].values[0]
        raw.at[index, '카테고리'] = df_row['알라딘카테고리명'].values[0].split(">")
    except IndexError:
        raw.drop(index, inplace=True)

raw
```

```python
raw.reset_index(drop=True, inplace=True)

raw.to_csv("data3.csv", index=True, encoding="utf-8",quoting=csv.QUOTE_ALL,escapechar="\\")
```

```python
df = pd.read_csv("data3.csv", quoting=csv.QUOTE_ALL, escapechar="\\", encoding="utf-8")

df['isbn'] = df['isbn'].astype(str)
df['year'] = df['year'].astype(int)
df['image_url'] = df['image_url'].astype(str)

def to_pg_array(s):
    s = s.replace("{", "[")
    s = s.replace("}", "]")
    s = s.split(",")
    s = [item.replace("'", '"') for item in s]
    return "{" + ",".join(f"{item}" for item in s) + "}"

df['category'] = df['category'].apply(lambda x: to_pg_array(x))

df['category']
```
