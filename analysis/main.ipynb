```python
import pandas as pd
```

```python
df = pd.read_csv('data3.csv')
```

```python
df['category'] = df['category'].apply(lambda x: eval(x)[1:])
df['category']
```

```python
import csv
df['category_depth'] = df['category'].apply(lambda x: len(x))
df['category_depth'].value_counts()
del df['category_depth']
df.to_csv("data3.csv", index=False, encoding="utf-8",quoting=csv.QUOTE_ALL,escapechar="\\")
```

```python


# 각 카테고리 경로의 모든 요소들을 하나의 집합으로 만들어 중복 제거
unique_category_elements = set()
for category_path in df['알라딘카테고리명']:
    unique_category_elements.update(category_path)

print(f"\n고유한 카테고리 요소의 수: {len(unique_category_elements)}")
print("\n고유한 카테고리 요소들:")
print(sorted(unique_category_elements))


```

```python
# 각 카테고리 요소별 등장 횟수를 계산
category_counts = {}
for category_path in df['알라딘카테고리명']:
    for category in category_path:
        category_counts[category] = category_counts.get(category, 0) + 1

# 등장 횟수를 기준으로 내림차순 정렬
sorted_category_counts = dict(sorted(category_counts.items(), key=lambda x: x[1], reverse=True))

print("\n각 카테고리 요소별 등장 횟수:")
for category, count in sorted_category_counts.items():
    print(f"{category}, {count}")
```
